## Preprocessing
### Normalization

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
feature = scaler.fit_transform(feature)
```

## Model Training

### K- Nearest Neighbour

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(feature,label,
                                                    test_size=0.2,
                                                    shuffle=True,
                                                    random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
i=0
while True:
    i+=1
    knn = KNeighborsClassifier(n_neighbors=i, weights='distance')
    knn.fit(x_train,y_train)
    score = knn.score(x_test,y_test)
    if score<1.0:
        i=i-1
        knn = KNeighborsClassifier(n_neighbors=i, weights='distance')
        knn.fit(x_train,y_train)
        score = knn.score(x_test,y_test)
        break
        

print(f"When n_neighbour is {i}, prediction is {score}")     
```

In the above cell, I found that any value that is less and equal to 375 will give me a score of 100% or 1.0.

```python
y_pred = knn.predict(x_test)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.metrics import classification_report, confusion_matrix

#confucion matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
# classificaton report
cr= classification_report(y_test,y_pred)
print(cr)

cm_display = ConfusionMatrixDisplay(confusion_matrix=cm)
cm_display.plot()
plt.show
```

### SVM

```python
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV

# Fold method to find the best parameter for the model.

param = {'C': [0.01, 0.1, 1, 10, 100],
         'gamma': [0.001, 0.01, 0.1, 1, 10, 100]}

SVC_grid = GridSearchCV(SVC(kernel='rbf'),
                       param_grid=param,
                       refit=True,
                       verbose=3)

SVC_grid.fit(x_train, y_train)

best_param = SVC_grid.best_params_
best_score = SVC_grid.best_score_
print(best_param)
print(best_score)
```

```python
clf_svm = SVC(kernel = 'rbf',gamma = best_param['gamma'], C=best_param['C'])
clf_svm.fit(x_train,y_train)
clf_svm.score(x_test,y_test)
```

```python
y_pred = clf_svm.predict(x_test)
```

```python
#confucion matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
# classificaton report
cr= classification_report(y_test,y_pred)
print(cr)

cm_display = ConfusionMatrixDisplay(confusion_matrix=cm)
cm_display.plot()
plt.show
```

### XGBoost

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score

xgb = XGBClassifier()
xgb.fit(x_train,y_train)

xgb_pred=xgb.predict(x_test)

score = accuracy_score(y_test, xgb_pred)
print(score)
```

```python
#confusion matrix
cm = confusion_matrix(y_test, xgb_pred)
print(cm)
# classificaton report
cr= classification_report(y_test,xgb_pred)
print(cr)

cm_display = ConfusionMatrixDisplay(confusion_matrix=cm)
cm_display.plot()
plt.show
```
