### Correlation
Statistical measure that defines to which extent two variables are linearly related to each other.
- Postive Correlation
- Zero Correlation
- Negative Correlation

### Pearson's Correlation
$$ r=\frac{\sum(x-m_{x})(y-m_{y})}{\sqrt{\sum(x-m_{x})^{2}\sum(y-m_{y})^{2}}} $$

#### Understanding Pearson's Correlation Coefficient

##### What is it?
Pearson's correlation coefficient (denoted as **r**) is a measure of the strength and direction of the relationship between two variables. It tells us how strongly two variables are related to each other.

##### How does it work?
- **Range**: The value of **r** ranges from **-1** to **1**.
  - **r = 1**: Perfect positive correlation (as one variable increases, the other also increases).
  - **r = -1**: Perfect negative correlation (as one variable increases, the other decreases).
  - **r = 0**: No correlation (the variables do not have any linear relationship).

##### Example
Imagine you're looking at the relationship between the hours studied and the marks obtained by students:
- If students who study more hours generally score higher, **r** will be close to **1**.
- If students who study more hours generally score lower, **r** will be close to **-1**.
- If there's no clear pattern between hours studied and marks, **r** will be close to **0**.

##### Why is it useful?
Pearson's correlation helps in understanding the strength of a relationship between two variables, which is useful in fields like data analysis, research, and statistics.

##### Important Note
- Pearson's correlation only measures linear relationships. Non-linear relationships won't be well represented by **r**.

2. Preprocessing

```python
# To do log transform to correct the skewed data.
import numpy as np

def log_transform(col):
    return np.log(col[0])

df['residual sugar'] = data['residual sugar'].apply(log_transform, axis=1)
df['chlorides'] = data['chlorides'].apply(log_transform, axis=1)
df['free sulfur dioxide'] = data['free sulfur dioxide'].apply(log_transform, axis=1)
df['total sulfur dioxide'] = data['total sulfur dioxide'].apply(log_transform, axis=1)
df['sulphates'] = data['sulphates'].apply(log_transform, axis=1)
```

```python
from sklearn.utils import resample

#Upsampling
quality_3_upsampled = resample(data[data['quality']==3],replace=True, n_samples=400, random_state=12)
quality_4_upsampled = resample(data[data['quality']==4],replace=True, n_samples=400, random_state=12)
quality_7_upsampled = resample(data[data['quality']==7],replace=True, n_samples=400, random_state=12)
quality_8_upsampled = resample(data[data['quality']==8],replace=True, n_samples=400, random_state=12)
```

```python
#downsampling
quality_5_downsampled = data[data['quality']==5].sample(n=400).reset_index(drop=True)
quality_6_downsampled = data[data['quality']==6].sample(n=400).reset_index(drop=True)
```

```python
balanced_data = pd.concat([quality_3_upsampled,
                           quality_4_upsampled,
                           quality_5_downsampled,
                           quality_6_downsampled,
                           quality_7_upsampled,
                           quality_8_upsampled]).reset_index(drop=True)
```

```python
balanced_data['quality'].value_counts()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set(style='whitegrid')
print(data['quality'].value_counts())
fig=plt.figure(figsize=(10,6))
sns.countplot(balanced_data,x='quality', palette='pastel')
```

```python
balanced_data.columns
```

```python
#Extracted features and label from dataframe
features = balanced_data[['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar',
       'chlorides', 'free sulfur dioxide', 'total sulfur dioxide', 'density',
       'pH', 'sulphates', 'alcohol']]
label = balanced_data['quality']
```

```python
# Normalization

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

x_scaled = scaler.fit_transform(features)
```

```python
# Label encoder

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y = le.fit_transform(label)
```

```python
# To check the label and the original quality index
le.inverse_transform([0,1,2,3,4,5])
```

**3. Model Training**

```python
# Train test split

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x_scaled,y,
                                                    test_size=0.2,
                                                    shuffle=True,
                                                    random_state=42)
```

```python
# Model training 
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(x_train,y_train)

score = knn.score(x_test,y_test)

print(score)
```

```python
# To find the optimal value of n_neighbors
for i in [3,5,7,9,11]:
    knn = KNeighborsClassifier(n_neighbors=i, weights='distance')
    knn.fit(x_train,y_train)
    score = knn.score(x_test,y_test)

    print(f"When n_neighbour is {i}, prediction is {score}")
```

```python
# Retrain the model base on the parameters above with best accuracy.

knn = KNeighborsClassifier(n_neighbors=7, weights='distance')
knn.fit(x_train,y_train)
score = knn.score(x_test,y_test)
print(score)
```

```python
y_pred = knn.predict(x_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

#confucion matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
# classificaton report
cr= classification_report(y_test,y_pred)
print(cr)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

cm_display = ConfusionMatrixDisplay(confusion_matrix=cm)
cm_display.plot()
plt.show
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV

param = {'C': [0.01, 0.1, 1, 10, 100],
         'gamma': [0.001, 0.01, 0.1, 1, 10, 100]}

SVC_grid = GridSearchCV(SVC(kernel='rbf'),
                       param_grid=param,
                       refit=True,
                       verbose=3)

SVC_grid.fit(x_train, y_train)
```

```python
best_param = SVC_grid.best_params_
best_score = SVC_grid.best_score_
print(best_param)
print(best_score)
```

```python
clf = SVC(kernel = 'rbf',gamma = best_param['gamma'], C=best_param['C'])
clf.fit(x_train,y_train)
clf.score(x_test,y_test)
```

```python
y_pred = clf.predict(x_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

#confucion matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
# classificaton report
cr= classification_report(y_test,y_pred)
print(cr)

cm_display = ConfusionMatrixDisplay(confusion_matrix=cm)
cm_display.plot()
plt.show
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score

xgb = XGBClassifier()
xgb.fit(x_train,y_train)

xgb_pred=xgb.predict(x_test)

score = accuracy_score(y_test, xgb_pred)
print(score)
```

```python
#confucion matrix
cm = confusion_matrix(y_test, xgb_pred)
print(cm)
# classificaton report
cr= classification_report(y_test,xgb_pred)
print(cr)

cm_display = ConfusionMatrixDisplay(confusion_matrix=cm)
cm_display.plot()
plt.show
```
