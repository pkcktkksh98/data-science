# Time Series Data (TSD)

- Time Series Data:
    - consist of data points
    - consistent time intervals
    - sequential
    - Component of TSD
        - Trends
        - Seasonality
        - Cyclical patterns
        - Irregular Variation
- Forecasting

```python
import numpy as np
import matplotlib.pyplot as plt

# Horizontal trend

t = np.arange(0,10,0.1)
data = np.zeros(len(t))+np.random.normal(0,0.1,len(t))
plt.plot(t,data,label='Horizontal Trend')

t = np.arange(0,10,0.1)
data = t+np.random.normal(0,0.5,len(t))
plt.plot(t,data,label='Upward Trend')

t = np.arange(0,10,0.1)
data = -t+np.random.normal(0,0.5,len(t))
plt.plot(t,data,label='Downward Trend')

# Damped Trend
t = np.arange(0, 10, 0.1)
data = np.exp(-0.1 * t) * np.sin(2 * np.pi * t) + np.random.normal(0, 0.5, len(t))
plt.plot(t, data, label='Damped Trend')

# Non-Linear Trend
t = np.arange(0, 10, 0.1)
data = t**2 + np.random.normal(0, 0.5, len(t))
plt.plot(t, data, label='Non-linear Trend')

plt.legend()
plt.show
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
time = np.arange(0, 366)

# Weekly seasonality
weekly_seasonality = np.sin(2 * np.pi * time / 7)
weekly_data = 5 + weekly_seasonality

# Monthly seasonality
monthly_seasonality = np.cos(2 * np.pi * time / 30)
monthly_data = 5 + monthly_seasonality

# Annual seasonality
annual_seasonality = np.sin(2 * np.pi * time / 365)
annual_data = 5 + annual_seasonality

# Plot the data
plt.figure(figsize=(12, 8))
plt.plot(time, weekly_data, label='Weekly Seasonality')
plt.plot(time, monthly_data, label='Monthly Seasonality')
plt.plot(time, annual_data, label='Annual Seasonality')
plt.legend(loc='upper left')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate sample data with cyclic patterns
np.random.seed(1)
time = np.array([0, 30, 60, 90, 120, 150, 180, 210, 240, 270, 300, 330])

data = 10 + 28 * np.sin(2 * np.pi * time / 58) + 20 * np.sin(2 * np.pi * time / 100)

# Plot the data
plt.figure(figsize=(12, 8))
plt.plot(time, data, label='Cyclic Data')
plt.legend(loc='upper left')
plt.xlabel('Time (days)')
plt.ylabel('Value')
plt.title('Cyclic Time Series Data')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)

time = np.arange(0, 100)
data =  10*np.sin(2 * np.pi * time / 20)+2*time

# Introduce irregularities by adding random noise
irregularities = np.random.normal(0, 25, len(data))
irregular_data = data + irregularities

# Plot the original data and the data with irregularities
plt.figure(figsize=(12, 8))
plt.plot(time, data, label='Original Data')
plt.plot(time, irregular_data, label='Data with Irregularities')
plt.legend(loc='upper left')
plt.show()
```

### Challenges

**Trend**: 
    - Identifying and accounting for long-term movements
**Seasonality**:
    - Recognizing and modeling periodic fluctuation that repeat at regular intervals
**Noise**:
    - Time series data often includes random fluctuations that can obscure underlying patternds
**irregularities**: 
    - Unexpected anomalies or outliers
**Missing Data**:
    - Filling or interpolating missing values while preserving the integrity of the time series can be challenging.
**Stationarity**:
    - Transforming non-stationary data can be necessary but complex
**Overfitting**:
    - Balancing model complexity to avoid overfitting is crucial
**Underfitting**:
    - Poor prediction
**Model Selection**:
    - Selecting the appropriate model based on the data characteristic  and forecasting needs can be challenging.
**Seasonal Adjustment**:
    - Correctly accounting for and removing seasonal effects without distorting the underl;ying data can be complex
**Temporal dependencies**:
    - Modeling these dependencies accurately requires understanding and incorporating lag effects.
**Data Granularity**:
    - Resolution and frequency: Time series data can be recorded at different frequencies.
**Forecasting Horizon**:
    - Short-term forecast may be more accurate that long-term forecast, which can be influenced by more uncertainty.
**Seasonal and Trend Changes**:
    - Seasonality and trends can evolve over time.
**External Factors**:
    - Many time series are influenced by external factors or events.
**Computational Complexity**:
    - Scalability: Handling large volumes of time series data or performin complex computations can be computationally intensive and require efficient algorithm and resources.
**Model Interpretability**:
    - Some tiome series models may act as black boxes, making it difficutl to interpret how predicitions are generated.

### AutoRegressive Integrated Moving Average (ARIMA)

- Statistical method for forecasting time series data.
- Key assumptions:
    - time series was generated from an underlying ARIMA process.
    - parameters p, d, q must be appropuately specified
- Components of ARIMA:
    - Auto Regressive (AR) **p**
    - Integrated (I) **d** 
    - Moving Average (MA) **q**

```python
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller

df = pd.read_csv("../../dataset/test.csv")
df
```

```python
df['Date']=pd.to_datetime(df['Date'])
```

```python
#Transforming Date column to be index
df.set_index('Date',inplace=True)
df
```

```python
# To aggregate my dataset

average = df.groupby(df.index)['number_sold'].mean()
average
```

```python
# To detect stationary
def test_stationary(timeseries):
    result = adfuller(timeseries.dropna(), autolag='AIC')
    print('ADF Statistic:', result[0])
    print('p-value:',result[1])
    print('Critical Values:', result[4])
    if result[1]<= 0.05:
        print("The time series is stationary")
    else:

        print("The timeseries is not stationary")
```

```python
test_stationary(average)
```

```python
average_data_diff = average.diff().dropna()
test_stationary(average_data_diff)
```

```python
# To train ARIMA model
model = ARIMA(average_data_diff, order=(12,0,15))
model_fit=model.fit()
```

```python
# To forecast

forecast_diff = model_fit.forecast(steps=5)

last_value = average.iloc[-1]
forecast = [last_value + forecast_diff[0]]
for diff in forecast_diff[1:]:
    forecast.append(forecast[-1] + diff)

forecast = pd.Series(forecast,
                     index=pd.date_range(start=average.index[-1]+pd.Timedelta(days=1), periods=5))
```

```python
plt.figure(figsize=(14,8))
plt.plot(average.index,average,color='blue',linestyle='-',label='Average Number Sold')
plt.plot(forecast.index, forecast, color = 'red',linestyle='--',label='Forecast')
plt.xlabel('Date')
plt.ylabel('Average Number Sold')
plt.title('ARIMA Forecast of Average Number Sold')
plt.legend()
plt.grid(True)
plt.xticks(rotation = 5)
plt.tight_layout()
plt.show()
```

```python
import plotly.express as px
import pandas as pd

df_aircraft = pd.read_csv('../../dataset/AirPassengers.csv')
df_aircraft
```

```python
df_aircraft.columns
```

```python
df_aircraft['Month'] = pd.to_datetime(df_aircraft['Month'])
```

```python
def plot_passenger_volumes(df,y):
    fig = px.line(df, x='Month',y=y,
                  labels={'Month','Date'})
    fig.update_layout(template='simple_white',
                      font=dict(size=18),
                      title_text='Airline Passengers',
                      width=650,
                      title_x=0.5,
                      height=400)
    return fig.show()
```

```python
plot_passenger_volumes(df_aircraft,'#Passengers')
```

```python
from scipy.stats import boxcox

df_aircraft['passengers_boxcox'],lam = boxcox(df_aircraft['#Passengers'])

plot_passenger_volumes(df_aircraft,'#Passengers')
```

```python
from statsmodels.tsa.arima.model import ARIMA
from scipy.special import inv_boxcox
train = df_aircraft.iloc[:-int(len(df_aircraft)*0.2)]
test = df_aircraft.iloc[-int(len(df_aircraft)*0.2):]
model = ARIMA(train['passengers_boxcox'], order=(12,1,12)).fit()
boxcox_forecast = model.forecast(len(test))
forecasts = inv_boxcox(boxcox_forecast, lam)
```

```python
import plotly.graph_objects as go

def plot_forecasts(forecasts,title):
    fig=go.Figure()
    fig.add_trace(go.Scatter(x = train['Month'],y=train['#Passengers'],name='Train'))
    fig.add_trace(go.Scatter(x = test['Month'],y=test['#Passengers'],name='Test'))
    fig.add_trace(go.Scatter(x = test['Month'],y=forecasts,name='Forecast'))
    fig.update_layout(template='simple_white',
                      font=dict(size=18),
                      title_text=title,
                      width=650,
                      title_x=0.5,
                      height=400)
    
    return fig.show()

plot_forecasts(forecasts,'ARIMA')
```

```python
model = ARIMA(train['passengers_boxcox'], order=(10,1,10),
              seasonal_order=(1,1,1,12)).fit()
boxcox_forecast_seasonal = model.forecast(len(test))
forecast_seasonal = inv_boxcox(boxcox_forecast_seasonal,lam)
```

```python
plot_forecasts(forecast_seasonal,'S-ARIMA')
```

```python
import pandas as pd
from statsmodels.tsa.seasonal import seasonal_decompose
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../../dataset/test.csv", parse_dates=['Date'], dayfirst=True)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date',inplace=True)

df=df.resample('D').sum()
```

```python
df.columns
```

```python
decomposition = seasonal_decompose(df['number_sold'],model='additive')
```

```python
plt.figure(figsize=(12,8))
plt.subplot(4,1,1)
plt.plot(df['number_sold'],label='Original')
plt.title('Original Time Series')
plt.legend()

plt.subplot(4,1,2)
plt.plot(decomposition.trend,label='Trend')
plt.title('Trend Component')
plt.legend()

plt.subplot(4,1,3)
plt.plot(decomposition.seasonal,label='Seasonal')
plt.title('Seasonal Component')
plt.legend()

plt.subplot(4,1,4)
plt.plot(decomposition.resid,label='Residual')
plt.title('Residual Component')
plt.legend()
```

```python
from statsmodels.tsa.api import SARIMAX

model = SARIMAX(df['number_sold'],
                order=(1,1,1),
                seasonal_order=(1,1,1,7))

result = model.fit()
```

```python
forecast_periods = 100
forecast = result.get_forecast(steps=forecast_periods)
forecast_index = pd.date_range(start = df.index[-1] + pd.Timedelta(days=1),periods=forecast_periods)
forecast_mean = forecast.predicted_mean
forecast_conf_int = forecast.conf_int()
```

```python
forecast_df = pd.DataFrame({
    'Forecast':forecast_mean,
    'Lower CI':forecast_conf_int.iloc[:,0],
    'Upper CI':forecast_conf_int.iloc[:,1],
}, index=forecast_index)
```

```python
plt.figure(figsize=(10,6))
plt.plot(df['number_sold'], label='Original Time Series')
plt.plot(forecast_df.index,forecast_df['Forecast'],label='Forecast',color='red',linestyle='--')
plt.fill_between(forecast_df.index,forecast_df['Lower CI'],forecast_df['Upper CI'],color='red',alpha=0.3)
plt.title('Time Series Forecast with SARIMAX')
plt.xlabel('Date')
plt.ylabel('Number Sold')
plt.legend()
plt.show()
```
